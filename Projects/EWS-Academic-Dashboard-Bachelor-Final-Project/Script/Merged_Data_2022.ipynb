```python
import pandas as pd

# Load dataset
mahasiswa = pd.read_excel("/content/drive/MyDrive/AKADEMIK KULIAH/(99) TUGAS AKHIR/Data TA/Data Mahasiswa/Data Pribadi Mahasiswa 2022.xlsx")
ipk = pd.read_excel("/content/drive/MyDrive/AKADEMIK KULIAH/(99) TUGAS AKHIR/Data TA/IPK/IPK 2022.xlsx")
khs = pd.read_excel("/content/drive/MyDrive/AKADEMIK KULIAH/(99) TUGAS AKHIR/Data TA/KHS/Cleaned_KHS_2022.xlsx")

# Format NIM ke string
for df in [mahasiswa, ipk, khs]:
    df["NIM"] = df["NIM"].astype(str)

# Mapping nilai
nilai_mapping = {'A': 4, 'A-': 3.7, 'B+': 3.3, 'B': 3, 'B-': 2.7, 'C+': 2.3, 'C': 2, 'C-': 1.7, 'D': 1, 'E': 0}
khs["Skor"] = khs["Nilai Akhir"].map(nilai_mapping)

# Ambil nilai tertinggi untuk matkul yang sama
khs = khs.sort_values(by=["NIM", "Nama Mata Kuliah", "Skor"], ascending=[True, True, False])
khs = khs.drop_duplicates(subset=["NIM", "Nama Mata Kuliah"], keep="first")

# Gabungkan data
df_merged = pd.merge(mahasiswa, ipk, on="NIM", how="left")
df_final = pd.merge(df_merged, khs, on="NIM", how="left")

# Tetapkan batas
IPK_MIN = 2.75
RISK_GRADES = ['C-', 'D', 'E']
batas_sks = {1: 20, 2: 20, 3: 21, 4: 20, 5: 20}

# Hitung total SKS per mahasiswa per semester (tanpa matkul E)
sks_per_semester = khs[~khs["Nilai Akhir"].isin(['C-', 'D', 'E'])].groupby(["NIM", "Semester"])["SKS"].sum().reset_index()
sks_per_semester.rename(columns={"SKS": "Total SKS"}, inplace=True)

# Gabungkan ke data final
df_final = pd.merge(df_final, sks_per_semester, on=["NIM", "Semester"], how="left")

# Tambahkan status SKS
def status_sks(row):
    batas = batas_sks.get(row['Semester'], 0)
    if pd.isna(row['Total SKS']) or row['Total SKS'] == 0:
        return "Tidak Aktif"
    elif row['Total SKS'] < batas:
        return "SKS Kurang"
    else:
        return "Aktif"

df_final["Status SKS"] = df_final.apply(status_sks, axis=1)

# Hitung IPS per baris (satu kolom)
def hitung_ips_baris(row, df_khs):
    nim = row['NIM']
    semester = row['Semester']
    data = df_khs[(df_khs['NIM'] == nim) & (df_khs['Semester'] == semester)]
    if data.empty or data['SKS'].sum() == 0:
        return None
    return round((data['SKS'] * data['Skor']).sum() / data['SKS'].sum(), 2)

df_final["IPS"] = df_final.apply(lambda row: hitung_ips_baris(row, khs), axis=1)

# Dictionary peminatan minimal 1 matkul
peminatan_dict = {
    'E-Business': {
        'PENGANTAR BISNIS DAN MANAJEMEN',
        'PENGANTAR BISNIS DIGITAL',
        'SISTEM INFORMASI AKUNTANSI'
    },
    'Enterprise': {
        'JAMINAN KUALITAS SISTEM INFORMASI',
        'SISTEM INFORMASI SDM',
        'ENTERPRISE RESOURCE PLANNING'
    },
    'User Experience': {
        'DASAR-DASAR DESAIN',
        'PRESENTASI DAN KOMUNIKASI VISUAL',
        'MULTIMEDIA INTERAKTIF'
    }
}

# Fungsi deteksi peminatan (minimal 1 matkul dari list)
def deteksi_peminatan(nim, df_khs):

    mk_all = set(df_khs[df_khs['NIM'] == nim]['Nama Mata Kuliah'].str.upper())
    for label, matkul_set in peminatan_dict.items():
        if mk_all.intersection(matkul_set):
            return label
    return 'Belum Teridentifikasi'

df_final['Peminatan'] = df_final['NIM'].apply(lambda nim: deteksi_peminatan(nim, khs))

# Fungsi risiko lulus
def risiko_lulus(row):
    ipk_risiko = row['IPK Total'] < IPK_MIN
    nilai_risiko = row['Nilai Akhir'] in RISK_GRADES
    sks_risiko = row["Status SKS"] in ["SKS Kurang", "Tidak Aktif"]
    return "Berpotensi Terlambat" if ipk_risiko or nilai_risiko or sks_risiko else "Aman"

df_final['Risiko Lulus'] = df_final.apply(risiko_lulus, axis=1)

# Simpan hasil akhir
df_final.to_excel("/content/drive/MyDrive/AKADEMIK KULIAH/(99) TUGAS AKHIR/Data TA/EWS 2022.xlsx", index=False)

# Cek hasil
df_final[['NIM', 'Nama', 'Semester', 'IPS', 'Status SKS', 'Peminatan', 'Risiko Lulus']].drop_duplicates().head()
```
