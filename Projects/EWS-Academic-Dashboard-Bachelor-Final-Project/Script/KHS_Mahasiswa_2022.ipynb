```python
# Codingan Fix
import pandas as pd

df = pd.read_excel('/content/drive/MyDrive/AKADEMIK KULIAH/(99) TUGAS AKHIR/Data TA/KHS/KHS Mentah 2022.xlsx', header=None)

# List untuk menampung hasil akhir
final_data = []

# Loop untuk membaca setiap mahasiswa
i = 0
while i < len(df):
    if isinstance(df.iloc[i, 0], str) and "NIM" in df.iloc[i, 0]:  # Deteksi awal mahasiswa
        nim = df.iloc[i, 2] if pd.notna(df.iloc[i, 2]) else None
        nama = df.iloc[i + 1, 2] if pd.notna(df.iloc[i + 1, 2]) else None
        program_studi = df.iloc[i, 5] if pd.notna(df.iloc[i, 5]) else None
        pembimbing = df.iloc[i + 1, 5] if pd.notna(df.iloc[i + 1, 5]) else None

        # Mulai mencari semester dalam data mahasiswa ini
        j = i + 1  # Pindah ke baris berikutnya
        semester = None

        while j < len(df):
            if isinstance(df.iloc[j, 0], str) and "Semester" in df.iloc[j, 0]:  # Deteksi semester
                semester_text = df.iloc[j, 0].split()
                if len(semester_text) > 1 and semester_text[1].isdigit():
                    semester = int(semester_text[1])

            elif pd.notna(df.iloc[j, 1]) and df.iloc[j, 1] != "NAMA MATA KULIAH":  # Data mata kuliah
                nama_mk = df.iloc[j, 1]
                sks = df.iloc[j, 3]
                nilai_akhir = df.iloc[j, 4]

                # Pastikan semua informasi penting tersedia sebelum menyimpan
                if nim and nama and semester and nama_mk:
                    final_data.append([nim, nama, program_studi, pembimbing, semester, nama_mk, sks, nilai_akhir])

            # Jika menemukan mahasiswa baru, berhenti loop semester
            if isinstance(df.iloc[j, 0], str) and "NIM" in df.iloc[j, 0]:
                break

            j += 1  # Pindah ke baris berikutnya

        i = j  # Lanjut ke mahasiswa berikutnya
    else:
        i += 1

# Buat DataFrame hasilnya
columns = ["NIM", "Nama", "Program Studi", "Pembimbing Akademik", "Semester", "Nama Mata Kuliah", "SKS", "Nilai Akhir"]
cleaned_df = pd.DataFrame(final_data, columns=columns)

# Hapus data kosong di Mata Kuliah dan Nilai Akhir
cleaned_df = cleaned_df.dropna(subset=["Nama Mata Kuliah", "Nilai Akhir"])

# Hapus baris yang memiliki ":" di Nama Mata Kuliah atau Nilai Akhir
cleaned_df = cleaned_df[~cleaned_df["Nama Mata Kuliah"].str.contains(":", na=False)]
cleaned_df = cleaned_df[~cleaned_df["Nilai Akhir"].astype(str).str.contains(":", na=False)]

# Hapus duplikat berdasarkan NIM, Mata Kuliah, dan Semester
cleaned_df = cleaned_df.drop_duplicates(subset=["NIM", "Nama Mata Kuliah", "Semester"])

# Tampilkan hasil bersih
cleaned_df
```

```python
# Simpan ke file Excel
cleaned_df.to_excel("Cleaned_KHS_2022.xlsx", index=False)
```
